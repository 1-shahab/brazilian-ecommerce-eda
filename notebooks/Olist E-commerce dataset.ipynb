# Handling DataTime

```python
orders["order_delivered_carrier_date"] = pd.to_datetime(orders["order_delivered_carrier_date"])
orders["order_delivered_customer_date"] = pd.to_datetime(orders["order_delivered_customer_date"])
orders["order_estimated_delivery_date"] = pd.to_datetime(orders["order_estimated_delivery_date"])
orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])
orders["order_approved_at"] = pd.to_datetime(orders["order_approved_at"])
```

```python
items["shipping_limit_date"] = pd.to_datetime(items["shipping_limit_date"])
```

# Merging data

```python
df_merged = pd.merge(orders,items,on='order_id',how='inner')
```

```python
df_final = pd.merge(df_merged,products,on='product_id',how='inner')
```

```python
df_final.head(3)
```

# EDA

## Time-Series Analysis

```python
 df_final["order_month"] = df_final["order_purchase_timestamp"].dt.to_period('M')
```

```python
df_final.head(3)
```

```python
monthly_orders = df_final.groupby("order_month").size()
monthly_orders
```

```python
plt.figure(figsize=(12,6))
monthly_orders.plot(kind='line')
plt.xlabel("Month")
plt.ylabel("count of orders")
plt.title("Monthly Orders")
plt.show()
```

## Prodyct & Categories Analysis

```python
top_products = df_final.groupby("product_category_name").size().sort_values().tail(10)
top_products
```

```python
plt.figure(figsize=(12,6))
top_products.plot(kind="barh")
plt.xlabel("count of sales")
plt.ylabel("product")
plt.title("Top Products")
plt.show()
```

## Price & Frieght Analysis

```python
x = df_final["price"]
y = df_final["freight_value"]
plt.scatter(x,y,alpha=0.5) 
plt.xlabel("Price")
plt.ylabel("Freight Value")
plt.title("Price & Frieght Analysis")
plt.show()
```
